أساسيات البرمجة الكائنية

```python
# البرمجة الكائنية (OOP) تبني قوالب (Classes) تجمع البيانات والوظائف معاً

class SimpleModel:
    def __init__(self, name):
        self.name = name  # خاصية (Attribute)
        self.is_trained = False

    def train(self):
        """وظيفة (Method) تغير حالة الكائن"""
        self.is_trained = True
        print(f" ⚙️ تم تدريب النموذج [{self.name}] بنجاح خلف الكواليس.")

# إنشاء كائن (Object) وتشغيله
my_model = SimpleModel("Linear_Regression")
print(f"حالة التدريب قبل: {my_model.is_trained}")
my_model.train()
print(f"حالة التدريب بعد: {my_model.is_trained}")
```
