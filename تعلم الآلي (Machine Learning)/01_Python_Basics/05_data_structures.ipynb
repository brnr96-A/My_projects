القوائم والقواميس

```python
# القوائم لتخزين السلاسل، والقواميس لتخزين البيانات المفتاحية (Key-Value)

# قائمة (List)
model_names = ["Linear", "Logistic", "K-Means"]
print(f"النموذج الثاني في القائمة: {model_names[1]}")

# قاموس (Dictionary)
customer_profile = {
    "name": "Sara",
    "age": 28,
    "has_loan": False
}
print(f"عمر العميل المستخرج من القاموس: {customer_profile['age']} سنة")
```
