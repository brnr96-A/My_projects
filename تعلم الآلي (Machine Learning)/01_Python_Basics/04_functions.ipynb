الدوال وإعادة الاستخدام

```python
# الدوال تجمع الكود لتسهيل استدعائه بمرونة عبر كلمة def

def calculate_net_income(gross_salary, tax_rate=0.1):
    """دالة بسيطة تحسب صافي الدخل بعد استقطاع الضرائب"""
    tax = gross_salary * tax_rate
    return gross_salary - tax

# تجربة استدعاء الدالة بقيم مختلفة
print(f"صافي راتب 5000$: ${calculate_net_income(5000)}")
print(f"صافي راتب 8000$ بضريبة مخصصة (15%): ${calculate_net_income(8000, 0.15)}")
```
