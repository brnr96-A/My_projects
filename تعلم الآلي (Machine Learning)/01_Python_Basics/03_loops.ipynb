 الحلقات التكرارية

```python
# الحلقات تكرر الكود دون إعادة كتابته يدوياً

print("--- تكرار القائمة باستخدام For ---")
features = ["Age", "Salary", "Location"]
for feature in features:
    print(f" └── معالجة الميزة: {feature}")

print("\n--- العداد التنازلي باستخدام While ---")
epochs = 3
while epochs > 0:
    print(f"  └── متبقي {epochs} دورات تدريبية")
    epochs -= 1
```
