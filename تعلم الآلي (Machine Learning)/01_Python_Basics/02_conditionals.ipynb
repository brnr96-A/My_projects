الشروط واتخاذ القرار

```python
# الشروط توجه البرنامج لاتخاذ قرارات بناءً على القيم
score = 85

print(f"الدرجة الحالية: {score}")
if score >= 90:
    print(" └── التقييم: ممتاز (Excellent)")
elif score >= 75:
    print(" └── التقييم: جيد جداً (Very Good)")
else:
    print(" └── التقييم: بحاجة لتحسين (Needs Improvement)")
```
