```python
import matplotlib.pyplot as plt
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.pipeline import make_pipeline
```

# 1. توليد بيانات تتبع منحنى غير خطي (علاقة تربية سينوسية كمثال)


```python
np.random.seed(42)
X_poly = np.sort(5 * np.random.rand(80, 1), axis=0)
y_poly = np.sin(X_poly).ravel() + np.random.normal(0, 0.1, X_poly.shape[0])
```

# 2. استخدام خط إنتاج (Pipeline) لرفع الأبعاد إلى الدرجة الثانية وتدريب الخيار الخطي فوراً


```python
degree = 3
poly_reg = make_pipeline(PolynomialFeatures(degree), LinearRegression())
poly_reg.fit(X_poly, y_poly)
```

```python
print("\n--- Polynomial Regression (Degree 3) Results ---")
# حساب كفاءة الملاءمة على نفس المنحنى
print(f"R2 Score: {poly_reg.score(X_poly, y_poly):.4f}")
```
