```python
from sklearn.linear_model import Ridge

# استخدام نفس بيانات الانحدار الخطي مع إضافة ضجيج عالي جداً لإحداث Overfitting عمداً
np.random.seed(42)
X_noisy = np.random.rand(20, 10) # 10 ميزات لعشرين عينة فقط (بيانات غير كافية)
y_noisy = 5 * X_noisy[:, 0] + np.random.normal(0, 2, 20)

X_train, X_test, y_train, y_test = train_test_split(X_noisy, y_noisy, test_size=0.3, random_state=42)

# تدريب Ridge Regression بمعامل تنظيم alpha=1.0
ridge_model = Ridge(alpha=1.0)
ridge_model.fit(X_train, y_train)

print("\n--- Ridge Regression (L2) Results ---")
print(f"Ridge Weights: {ridge_model.coef_}")
print(f"Test Score (R2): {ridge_model.score(X_test, y_test):.4f}")
```
