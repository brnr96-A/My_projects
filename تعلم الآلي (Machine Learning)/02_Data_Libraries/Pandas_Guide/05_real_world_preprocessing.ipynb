```python
import pandas as pd
import numpy as np

# مصفوفة بيانات خام تحاكي بيانات العملاء في شركة
raw_dataset = {
    'Customer_ID': [1, 2, 3, 4],
    'Income': [45000, 60000, np.nan, 80000],
    'Gender': ['Male', 'Female', 'Female', 'Male'],
    'Purchased': ['No', 'Yes', 'No', 'Yes']
}
df_customers = pd.DataFrame(raw_dataset)

print("--- خطوة 1: معالجة القيم المفقودة في الدخل ---")
df_customers['Income'] = df_customers['Income'].fillna(df_customers['Income'].mean())

print("\n--- خطوة 2: تحويل البيانات الفئوية لنصوص (One-Hot Encoding) ---")
# خوارزميات الـ ML لا تفهم النصوص، نقوم بتحويل عمود Gender إلى أعمدة أرقام 0 و 1
df_customers = pd.get_dummies(df_customers, columns=['Gender'], drop_first=True, dtype=int)

print("\n--- خطوة 3: ترميز عمود الهدف (Label Encoding) ---")
# تحويل عمود Purchased (No -> 0, Yes -> 1)
df_customers['Purchased'] = df_customers['Purchased'].map({'No': 0, 'Yes': 1})

print("\n--- الجدول النهائي الجاهز للتمرير إلى NumPy ومصفوفة X, y ---")
print(df_customers)

# فصل المصفوفات وتحويلها لـ NumPy arrays تمهيداً للتدريب
X = df_customers.drop(columns=['Customer_ID', 'Purchased']).values
y = df_customers['Purchased'].values
print("\nشكل المصفوفة النهائية الجاهزة للتدريب في NumPy:", X.shape)
```
