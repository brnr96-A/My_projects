```python
import pandas as pd
```

```python
df = pd.DataFrame({
    'City': ['New York', 'Paris', 'Tokyo', 'Paris'],
    'Score':,
    'Verified': [True, False, True, True]
})
```

# 1. التصفية الشرطية (Filtering)

```python
# جلب البيانات التي تحقق شرط: المدينة Paris والتقييم أعلى من 80
paris_high_score = df[(df['City'] == 'Paris') & (df['Score'] > 80)]
```

# 2. استخدام loc (بناءً على التسميات والفهرس المنطقي)

```python
# جلب صفوف معينة وأعمدة محددة بالاسم
print(df.loc[df['Score'] > 80, ['City', 'Verified']])
```

# 3. استخدام iloc (بناءً على فهارس الأرقام الصرفة - يشبه NumPy)


```python
# جلب أول 3 صفوف مع أول عمودين
X_features = df.iloc[:3, :2]
```

# 4. الترتيب (Sorting)

```python
df_sorted = df.sort_values(by='Score', ascending=False) # ترتيب تنازلي حسب السكور
```
