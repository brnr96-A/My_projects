import pandas as pd
import numpy as np

```python
# جدول بيانات يحتوي على مشاكل (قيم مفقودة ومكررة)
raw_data = {
    'Feature_1': [10, 20, np.nan, 20, 50],
    'Feature_2': [5.5, np.nan, 7.2, np.nan, 9.1],
    'Label': ['Yes', 'No', 'Yes', 'No', 'No']
}
df = pd.DataFrame(raw_data)
```

# 1. اكتشاف القيم المفقودة (Null/NaN Values)

```python
print(df.isnull().sum()) # عدد القيم المفقودة في كل عمود
```

# 2. معالجة القيم المفقودة (Imputation)

```python
df_dropped = df.dropna() # الخيار أ: حذف أي سطر يحتوي على قيمة مفقودة
df['Feature_1'] = df['Feature_1'].fillna(df['Feature_1'].median()) # الخيار ب: تعبئة المفقود بالوسيط (شائع في الـ ML)
```

# 3. معالجة البيانات المكررة (Duplicate Data)

```python
print(df.duplicated().sum()) # معرفة الأسطر المكررة بالكامل
df = df.drop_duplicates().reset_index(drop=True) # حذف التكرار وإعادة ترتيب الفهرس
```
