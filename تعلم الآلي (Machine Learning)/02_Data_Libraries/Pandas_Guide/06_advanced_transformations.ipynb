```python
import pandas as pd
import numpy as np
```

```python
# جدول بيانات متقدم
df = pd.DataFrame({
    'Employee': ['Ahmed', 'Fatima', 'Zaid', 'Sara'],
    'Department': ['IT', 'HR', 'IT', 'Sales'],
    'Salary':,
    'Join_Date': ['2024-01-15', '2023-05-22', '2025-11-01', '2024-08-10']
})
```

# 1. دالة apply و lambda (لتطبيق عمليات مخصصة ومعقدة على الأعمدة)

```python
# زيادة الرواتب بنسبة 10% فقط لموظفي قسم الـ IT
df['New_Salary'] = df.apply(lambda row: row['Salary'] * 1.1 if row['Department'] == 'IT' else row['Salary'], axis=1)
```

# 2. دالة transform (مفيدة جداً لحساب ميزات نسبية بناءً على المجموعات)

```python
# حساب متوسط رواتب القسم وإلحاقه بكل موظف لمعرفة موقعه من متوسط قسمه
df['Dept_Avg_Salary'] = df.groupby('Department')['Salary'].transform('mean')
```

# 3. التعامل مع التواريخ والسلاسل الزمنية (Time-Series Basics)


```python
df['Join_Date'] = pd.to_datetime(df['Join_Date']) # تحويل النص إلى كائن تاريخ ذكي
df['Join_Year'] = df['Join_Date'].dt.year        # استخراج السنة كميزة منفصلة
df['Join_Month'] = df['Join_Date'].dt.month      # استخراج الشهر كميزة منفصلة
```
