```python
import pandas as pd
```

# 1. التجميع (GroupBy)

```python
df_sales = pd.DataFrame({
    'Department': ['Electronics', 'Clothing', 'Electronics', 'Clothing'],
    'Sales':,
    'Bonus': [100, 50, 80, 20]
})
# حساب مجموع المبيعات ومتوسط البونص لكل قسم
grouped = df_sales.groupby('Department').agg({'Sales': 'sum', 'Bonus': 'mean'})
print(grouped)
```

# 2. دمج الجداول (Merging / Joining) - يشبه روابط SQL

```python
df_users = pd.DataFrame({'User_ID':, 'Name': ['Omar', 'Sara', 'Zaid']})
df_orders = pd.DataFrame({'User_ID':, 'Product': ['Laptop', 'Phone', 'Book']})

# دمج داخلي (Inner Join) بناءً على المفتاح المشترك User_ID
df_merged = pd.merge(df_users, df_orders, on='User_ID', how='inner')
print(df_merged)
```
