```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd
```

```python
# توليد بيانات افتراضية (أعمار وتوزيع دخل)
np.random.seed(42)
ages = np.random.normal(loc=35, scale=10, size=500) # توزيع طبيعي للأعمار
income = np.random.exponential(scale=50000, size=500) # توزيع ملتوي للدخل

df = pd.DataFrame({'Age': ages, 'Income': income})
```

```python
# إعداد شكل الرسم البياني عبر Matplotlib
plt.figure(figsize=(12, 5))
```

# 1. رسم التوزيع الإحصائي (Histogram + KDE) باستخدام Seaborn

```python
plt.subplot(1, 2, 1) # تقسيم اللوحة (صف واحد، عمودين، الرسمة الأولى)
sns.histplot(df['Age'], kde=True, color='skyblue')
plt.title('Age Distribution (Gaussian Pattern)')
plt.xlabel('Age')
```

# 2. رسم الصندوق الإحصائي (Box Plot) لاكتشاف القيم الشاذة (Outliers)

```python
plt.subplot(1, 2, 2) # الرسمة الثانية
sns.boxplot(x=df['Income'], color='salmon')
plt.title('Income Box Plot (Detecting Outliers)')
plt.xlabel('Income')
```

```python
plt.tight_layout() # تجميع متناسق للرسومات بدون تداخل الهوامش
plt.show()
```
