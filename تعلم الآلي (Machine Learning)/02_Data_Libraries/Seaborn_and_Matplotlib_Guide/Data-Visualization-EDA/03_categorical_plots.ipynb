```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# بيانات مبيعات افتراضية للهواتف
df_products = pd.DataFrame({
    'Brand': ['Apple', 'Samsung', 'Google', 'Apple', 'Samsung', 'Google', 'Apple', 'Samsung'],
    'Rating': [4.8, 4.5, 4.6, 4.7, 4.4, 4.5, 4.9, 4.3],
    'Region': ['US', 'US', 'US', 'EU', 'EU', 'EU', 'AS', 'AS']
})
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
```


# 1. مخطط الأعمدة (Bar Plot) - يحسب المتوسط تلقائياً مع هامش الخطأ الإحصائي

```python
sns.barplot(data=df_products, x='Brand', y='Rating', ax=axes[0], palette='muted')
axes[0].set_title('Average Rating per Brand')
```

# 2. مخطط الكمان (Violin Plot) - يدمج بين الـ Box Plot والـ KDE لرؤية الكثافة الاحتمالية للفئات

```python
sns.violinplot(data=df_products, x='Brand', y='Rating', hue='Region', ax=axes[1], split=True, palette='pastel')
axes[1].set_title('Rating Distribution Density per Brand & Region')

plt.show()
```
