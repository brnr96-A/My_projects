```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
```

```python
# إنشاء مصفوفة ميزات معقدة لـ 6 ميزات مرتبطة ببعضها
np.random.seed(42)
data = np.random.rand(100, 5)
df_features = pd.DataFrame(data, columns=['Feature_A', 'Feature_B', 'Feature_C', 'Feature_D', 'Target'])

# جعل الميزة B مرتبطة جداً بالميزة A عمداً (تكرار معلومة)
df_features['Feature_B'] = df_features['Feature_A'] * 0.95 + np.random.normal(0, 0.05, 100)
```

# 1. حساب مصفوفة الارتباط (Correlation Matrix) باستخدام Pearson correlation


```python
corr_matrix = df_features.corr()
```

# 2. رسم الخريطة الحرارية (Heatmap) للارتباطات

```python
plt.figure(figsize=(8, 6))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5, vmin=-1, vmax=1)

plt.title('Feature Correlation Heatmap for Machine Learning')
plt.show()

print("""
💡 رؤية مهندس التعلم الآلي عند قراءة هذا المخطط:
- نلاحظ أن الارتباط بين Feature_A و Feature_B يساوي 0.95 (قريب جداً من 1).
- الإجراء الهندسي: يجب حذف إحدى الميزتين قبل البدء بالتدريب لمنع حدوث مشكلة الـ Multicollinearity وسرعة الحساب.
- نلاحظ ارتباط Feature_A مع الـ Target جيد، لذا نُبقي عليها.
""")
```
