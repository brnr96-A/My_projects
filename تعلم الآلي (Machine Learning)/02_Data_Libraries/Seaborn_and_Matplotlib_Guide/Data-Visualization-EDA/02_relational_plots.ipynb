```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
```

```python
# توليد بيانات تحاكي العلاقة بين مساحة المنزل وسعره
sqft = np.random.randint(800, 4000, 100)
price = sqft * 150 + np.random.normal(0, 30000, 100) # علاقة خطية مع ضجيج
house_type = np.random.choice(['Apartment', 'Villa'], 100)

df_houses = pd.DataFrame({'SqFt': sqft, 'Price': price, 'Type': house_type})
```

```python
# رسم مخطط التشتت (Scatter Plot) لدراسة العلاقة بين ميزتين رقميتين
plt.figure(figsize=(8, 6))
```

```python
# استخدام معامِل hue لتلوين النقاط بناءً على فئة نوع المنزل (ميزة جبارة في Seaborn)
sns.scatterplot(data=df_houses, x='SqFt', y='Price', hue='Type', palette='viridis', alpha=0.8)
```

```python
# رسم خط الاتجاه الخطي الافتراضي (Regression Line) باستخدام sns.regplot لـ Matplotlib كواليس
sns.regplot(data=df_houses, x='SqFt', y='Price', scatter=False, color='red', label='Trend Line')

plt.title('House Price vs Square Footage')
plt.xlabel('Square Feet (Feature X)')
plt.ylabel('Price (Target Y)')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5) # إضافة شبكة خلفية باستخدام Matplotlib
plt.show()
```
