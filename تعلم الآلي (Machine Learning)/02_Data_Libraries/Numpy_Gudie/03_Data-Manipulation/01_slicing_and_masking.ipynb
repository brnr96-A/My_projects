# استدعاء مكتبة ال NUMPY

```python
import numpy as np
```

# مصفوفة بيانات افتراضية تحاكي جدول ميزات وعمود الهدف الأخير

```python
dataset = np.array([[25, 50, 1],[12, 20, 0]])
```

# 1. فصل الميزات (X) عن الهدف (y)

```python
X = dataset[:, :-1] # كل الصفوف، وكل الأعمدة ما عدا الأخير
y = dataset[:, -1]  # كل الصفوف، العمود الأخير فقط

print("Features X:\n", X)
print("Target y:", y)
```

# 2. التصفية الشرطية (Boolean Masking) لتنظيف البيانات الشاذة

```python
outliers = np.array([15, 200, 30, -5, 45])
valid_mask = (outliers >= 0) & (outliers <= 100) # قناع منطقي
cleaned_data = outliers[valid_mask] # استخلاص العناصر الصحيحة فقط
print("Cleaned Data:", cleaned_data)
```

# 3. التعديل الشرطي الفوري

```python
outliers[outliers < 0] = 0 # استبدال القيم السالبة بـ 0
```
