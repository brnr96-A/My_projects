# 2. مفهوم البث (Broadcasting)

```python
# مصفوفة (3x2) مع متجه (1x2)
X_data = np.array([[10, 100], [20, 200], [30, 300]])
mean_vector = np.array([20, 200])

# يتم طرح المتجه من كل صف في المصفوفة تلقائياً عبر البث دون تكرار الذاكرة
normalized_X = X_data - mean_vector
print("Normalized Data:\n", normalized_X)
```
