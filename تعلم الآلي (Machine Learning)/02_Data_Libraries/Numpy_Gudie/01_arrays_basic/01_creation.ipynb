# إنشاء المصفوفات من الصفر بعدة طرق

# استدعاء مكتبة Numpy

```python
import numpy as np
```

# إنشاء مصفوفة من قائمة بايثون عادية

```python
list_data = [1, 2, 3, 4, 5]
arr = np.array(list_data)
print(f"المصفوفة: {arr}\n")
```

# إنشاء مصفوفة من أصفار و مصفوفة أخرى من آحاد بأبعاد مختلفة

```python
zeros_arr = np.zeros((2, 3))  # مصفوفة بأبعاد 2*3 مليئة بالأصفار (لتصفير الأوزان الابتدائية)
ones_arr = np.ones((3, 3))    # مصفوفة بأبعاد 3*3 مليئة بالآحاد
print(f"الأصفار:\n{zeros_arr}")
print(f"الآحاد:\n{ones_arr}\n")
```

# إنشاء مصفوفة ذات تسلسل رقمي منظم 

```python
range_arr = np.arange(0, 10, 2)  # توليد أرقام من 0 إلى أقل من 10 بخطوة 2
print(f"التسلسل الرقمي: {range_arr}")
```
