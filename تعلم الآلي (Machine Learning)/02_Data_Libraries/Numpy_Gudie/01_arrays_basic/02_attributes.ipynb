#  خصائص المصفوفات (Array Attributes)

```python
# استدعاء مكتبة ال Numpy 
```

```python
import numpy as np
```

# مصفوفة ثنائية الأبعاد (جدول من صفين و 3 أعمدة) تمثل ميزات عينة بيانات

```python
arr = np.array([[1, 2, 3], [4, 5, 6]])

print(f"المصفوفة المستهدفة:\n{arr}\n")
```

# لمعرفة عدد الأبعاد (مهم جداً للتأكد من توافق البيانات مع شروط خوارزميات الـ ML)


```python
print(f"عدد الأبعاد (ndim): {arr.ndim}")
```

# لمعرفة الشكل الهيكلي (عدد الصفوف والأعمدة)

```python
print(f"الشكل الهيكلي (shape): {arr.shape}")
```

# لمعرفة نوع البيانات المخزنة بالداخل

```python
print(f"نوع البيانات (dtype): {arr.dtype}")
```


# لمعرفة حجم المصفوفة

```python
print(f"العدد الإجمالي للعناصر (size): {arr.size}")
```
