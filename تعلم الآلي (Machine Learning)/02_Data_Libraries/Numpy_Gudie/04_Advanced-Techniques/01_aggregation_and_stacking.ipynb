# استدعاء المكتبة NUMPY

```python
import numpy as np
```

```python
loss_matrix = np.array([[0.5, 0.1],
                        [0.2, 0.4],
                        [0.1, 0.3]])
```

# 1. الحسابات عبر المحاور (Axes)

```python
# axis=0: العمل عمودياً (حساب المتوسط لكل ميزة عبر العينات)
print("Mean per feature (axis=0):", np.mean(loss_matrix, axis=0))
# axis=1: العمل أفقياً (حساب مجموع الأخطاء لكل عينة بشكل منفصل)
print("Sum per sample (axis=1):", np.sum(loss_matrix, axis=1))
```

# 2. الدمج والتكديس (Concatenation & Stacking)

```python
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

concat_v = np.vstack((A, B)) # تكديس عمودي (إضافة عينات جديدة)
concat_h = np.hstack((A, B)) # تكديس أفقي (إضافة ميزات جديدة)
```

# 3. حيلة خدعة عمود الانحياز (The Bias Column Trick) في خوارزميات الـ ML

```python
X_orig = np.array([[10, 20], [30, 40], [50, 60]])
# دمج عمود مليء بالواحدات (1s) في مقدمة المصفوفة ليمثل قيمة الـ Bias رياضياً
X_with_bias = np.c_[np.ones(X_orig.shape[0]), X_orig]
print("X with Bias Column:\n", X_with_bias)
```
