# 2. العمليات الإحصائية (Statistical Functions)

# استدعاء المكتبة Numpy

```python
import numpy as np
```

# علامات لمجموعة من الطلاب (صفين يمثلان طالبين، و 3 أعمدة تمثل 3 مواد)

```python
grades = np.array([[85, 90, 78], [92, 88, 95]])
```

# طباعة مصفوفة العلامات و المواد 

# حساب المجموع الكلي لكل العلامات
# حساب المتوسط الحسابي 
# حساب أعلى درجة 


```python
print(f"مصفوفة الدرجات الأصلية:\n{grades}\n")
print(f"المجموع الكلي للدرجات: {np.sum(grades)}")
print(f"المتوسط الحسابي العام لجميع البيانات (Mean): {np.mean(grades):.2f}")
print(f"أعلى درجة مسجلة (Max): {np.max(grades)}")
```

# -- الحساب الإحصائي على مستوى المحاور (Axes) --
# axis=0 يعني الحساب عمودياً (لكل مادة)، axis=1 يعني الحساب أفقياً (لكل طالب)


```python
print(f"المتوسط لكل مادة على حدة (عمودياً axis=0): {np.mean(grades, axis=0)}")
print(f"المتوسط لكل طالب على حدة (أفقياً axis=1): {np.mean(grades, axis=1)}")
```
