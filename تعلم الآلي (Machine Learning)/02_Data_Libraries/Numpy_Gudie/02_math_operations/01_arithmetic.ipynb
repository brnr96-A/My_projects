# 1. العمليات الحسابية (Element-wise & Matrix Arithmetic)

# استدعاء مكتبة Numpy

```python
import numpy as np
```

# تعريف مصفوفتين a , b

```python
a = np.array([10, 20, 30])
b = np.array([1, 2, 3])
```

العمليات العنصرية (Element-wise) 

```python
print(f"الجمع العنصري (a + b): {a + b}")
print(f"الطرح العنصري (a - b): {a - b}")
print(f"الضرب العنصري (a * b): {a * b}\n")
```

# --- ضرب المصفوفات الرياضي الحقيقي (Dot Product) ---
# ضرب مصفوفة الأوزان بمصفوفة المدخلات في الشبكات العصبية والـ Linear Regression


```python
matrix_x = np.array([[1, 2], [3, 4]])
matrix_y = np.array([[5, 6], [7, 8]])
```

# حساب الضرب الحقيقي للمصفوفات 


```python
result = np.dot(matrix_x, matrix_y)
```
