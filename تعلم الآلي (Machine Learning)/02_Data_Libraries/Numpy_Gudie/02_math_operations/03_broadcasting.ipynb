# 3. البث (Broadcasting)
# ميزة تتيح إجراء العمليات الحسابية على مصفوفات بأبعاد مختلفة دون الحاجة لنسخ أو تكرار البيانات يدوياً

# استدعاء المكتبة Numpy

```python
import numpy as np
```

# تعريف ماتريس


```python

matrix = np.array([[1, 2, 3], 
                   [4, 5, 6]])
```

# إضافة رقم مفرد (Scalar) للمصفوفة بأكملها
# سيقوم NumPy بـ "بث" الرقم 10 تلقائياً ليتناسب مع أبعاد المصفوفة ويضاف لكل عنصر

```python
result = matrix + 10

print("--- البث لرقم مفرد ---")
print(f"المصفوفة بعد إضافة 10 لكل عنصر عبر البث:\n{result}\n")
```

# --- البث لمتجه (Vector) على مصفوفة ---
  # مصفوفة أحادية ذات 3 عناصر متوافقة مع عدد الأعمدة

```python
vector = np.array([1, 0, -1])
```


# طباعة المصفوفة النهائية بعد إضافة المتجه

```python
result_vector = matrix + vector
print(f"المصفوفة بعد إضافة المتجه لكل صف تلقائياً:\n{result_vector}")
```
