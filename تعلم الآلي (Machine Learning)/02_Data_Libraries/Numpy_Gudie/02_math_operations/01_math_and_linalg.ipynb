# 1. العمليات الحسابية (Element-wise & Matrix Arithmetic)

# استدعاء مكتبة Numpy

```python
import numpy as np
```

# تهيئة متجهات ميزات وأوزان افتراضية

```python
W = np.array([0.5, 1.2, -0.8])  # Weights
X = np.array([2.0, 3.0, 1.5])   # Features
```

# 1. العمليات عنصر بعنصر (Element-wise Operations)

```python
print("X + 2 =", X + 2)
print("X * W =", X * W) # ضرب كل عنصر بما يقابله
print("e^X =", np.exp(X)) # دالة الأس (تستخدم لاحقاً في خوارزمية الـ Sigmoid)
```

# 2. الضرب القياسي لمتجهين (Dot Product)

```python
# الرياضات: Z = (0.5*2.0) + (1.2*3.0) + (-0.8*1.5)
dot_product = np.dot(W, X)
print(f"Dot Product (Z): {dot_product}")
```

# 3. ضرب المصفوفات (Matrix Multiplication)

```python
Features = np.array([[1, 2], , 
                     [5, 6]])
# مصفوفة أوزان (2x1)
Weights = np.array([[0.1], 
                    [0.2]])
```

```python
# ضرب مصفوفة (3x2) في (2x1) ينتج مصفوفة مخرجات (3x1)
predictions = Features @ Weights
print("Predictions:\n", predictions)
```
