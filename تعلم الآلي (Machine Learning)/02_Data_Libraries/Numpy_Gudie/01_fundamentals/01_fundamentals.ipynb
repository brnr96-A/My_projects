```python
import numpy as np
```

# 1. إنشاء مصفوفات بأبعاد مختلفة ومصطلحات الـ ML

```python
scalar = np.array(5)                    # 0-D: كمية قياسية
vector = np.array([1.0, 2.0, 3.0])      # 1-D: متجه (مثال: أوزانخوارزمية)
matrix = np.array([[1, 2], [3, 4]])     # 2-D: مصفوفة (مثال: جدول ميزات العينات)
```

# 2. فحص الخصائص (مهم لتتبع أخطاء الأبعاد المدمجة)

```python
print(f"Shape: {matrix.shape}") # المخرجات: (2, 2)
print(f"Data Type: {matrix.dtype}") # النوع الداخلي
print(f"Total Elements: {matrix.size}")
```
